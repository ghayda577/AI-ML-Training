## Step 3: Detect Outliers in Monthly Income

A box plot was created to identify potential outliers in the Monthly Income variable.

Box plots provide a clear summary of the data distribution and highlight values that fall outside the expected range.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df["MonthlyIncome"])
plt.title("Monthly Income Outliers")
plt.show
```

### Observation

Several high-income values appear outside the upper whisker, indicating the presence of outliers in the Monthly Income feature.

## Step 4: Handle Monthly Income Outliers

The Interquartile Range (IQR) method was used to calculate the lower and upper bounds for detecting outliers.

Instead of removing these observations, outliers were capped by limiting values that exceeded the upper bound. This approach preserves all records while reducing the influence of extreme values.

```python
income=df["MonthlyIncome"]
Q1= income.quantile(0.25)
Q3= income.quantile(0.75)
IQR=Q3 - Q1
LB=Q1 - (1.5*IQR)
UB=Q3 +(1.5*IQR)
print("IQR =", IQR)
print("Lower Bound = ", LB)
print("Upper Bound =",UB)

```

```python
df["MonthlyIncome-capped"] =income.clip(
    lower=LB,
    upper=UB
)
plt.figure(figsize=(6,4))
sns.boxplot(x=df["MonthlyIncome-capped"])
plt.title("Monthly Income Outliers after capping")
plt.show
```

### Observation

After applying capping, the extreme Monthly Income values were successfully limited. The box plot shows a more balanced distribution with reduced impact from outliers while retaining all observations.

## Step 5: Analyze Relationships Between Variables

In this step, bivariate analysis was performed to explore the relationships between important variables in the dataset.

Different visualizations were created to investigate how numerical variables relate to each other and how employee characteristics differ across groups. These plots help identify patterns, trends, differences between categories, and possible factors related to employee attrition.

The following relationships were explored:
- Relationship between employee age and monthly income using a scatter plot.
- Comparison of monthly income across different job levels using grouped box plots.
- Comparison of age distribution based on employee attrition status.
- Comparison of monthly income distribution based on employee attrition status.

### 5.1 Relationship Between Age and Monthly Income

A scatter plot was created to examine the relationship between employee age and monthly income.

Scatter plots help identify whether a relationship exists between two numerical variables, as well as reveal trends, clusters, and potential outliers.

```python
plt.figure(figsize=(12,8))
sns.scatterplot(data=df , x="Age", y="MonthlyIncome")
plt.title("Relationship between Age and Monthly Income")
plt.show
```

### Observation

The scatter plot shows a slight positive relationship between Age and Monthly Income. Although older employees tend to have higher monthly incomes, the data points are widely scattered. This indicates that age alone does not strongly explain variations in monthly income, and other factors, such as job level or years of experience, are likely to have a greater influence.

### 5.2 Monthly Income Across Job Levels

A grouped box plot was created to compare the distribution of monthly income across different job levels.

This visualization helps identify how employee salaries vary by job level, while also showing the median, spread, and potential outliers within each group.

```python
plt.figure(figsize=(12,8))
sns.boxplot(data=df, x="JobLevel" , y="MonthlyIncome")
plt.title("Monthly Income Distribution by Job Level")
plt.show
```

### Observation

Monthly income generally increases as job level increases. Employees at higher job levels tend to earn significantly higher salaries than those at lower levels. The spread of monthly income also becomes larger at higher job levels, indicating greater salary variation among senior employees.

### 5.3 Age Distribution by Employee Attrition

A grouped box plot was created to compare the age distribution of employees based on their attrition status.

This visualization helps determine whether employee age differs between those who left the company and those who remained.

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x=df["Attrition"], y=df["Age"])
plt.title("Age Distribution by Employee Attrition")
```

### Observation

The age distributions of employees who stayed and those who left are relatively similar. However, employees who left the company tend to be slightly younger on average, while employees who remained have a slightly higher median age.

### 5.4 Monthly Income by Employee Attrition

A grouped box plot was created to compare the distribution of monthly income between employees who left the company and those who remained.

This visualization helps determine whether salary may be associated with employee attrition by comparing the median, spread, and potential outliers of each group.

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x="Attrition", y="MonthlyIncome")
plt.title("Monthly Income Distribution by Employee Attrition")
```

### Observation

Employees who remained at the company generally have higher monthly incomes than those who left. In addition, the "No" attrition group shows a wider range of salaries, while employees who left tend to have lower monthly incomes. This suggests that monthly income may be associated with employee retention.

## Step 6: Analyze Correlations Between Numerical Variables

In this step, a correlation analysis was performed to measure the relationships between numerical variables in the dataset.

The correlation matrix was calculated using Pearson correlation, which shows the strength and direction of the relationship between each pair of numerical variables.

The correlation values were first displayed as a matrix, then visualized using an annotated heatmap to make it easier to identify strong positive, negative, or weak relationships between variables.

```python
corr = df.corr(numeric_only=True)
print(corr)
```

### Correlation Heatmap Visualization

The heatmap provides a visual representation of the correlation matrix, where the color intensity represents the strength of the relationship between variables.

```python
plt.figure(figsize=(14,10))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Matrix of Numerical Features")
plt.show()
```
