## Previous Data Cleaning

During the previous phase, data quality issues were identified and
addressed.

In particular, zero values in `RestingBP` and `Cholesterol` were identified
as invalid values and treated as missing values.

Missing values were then handled through the preprocessing pipeline using
median imputation for numerical features and most-frequent imputation for
categorical features.

These validated cleaning decisions are reused in Phase 3.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/heart.csv")

df["RestingBP"] = df["RestingBP"].replace(0, np.nan)
df["Cholesterol"] = df["Cholesterol"].replace(0, np.nan)

print("Dataset shape:", df.shape)
```

## Previous Feature Engineering

Feature engineering was evaluated systematically during the previous
phase.

Two engineered features were investigated:

- `AgeGroup`
- `health_risk_count`

The experiments compared the original feature set with different
combinations of engineered features.

The results were used to determine which engineered feature should be
retained for the final workflow.

Based on the previous evaluation, `health_risk_count` was retained, while
`AgeGroup` was excluded from the final feature set.

Therefore, Phase 3 uses the validated feature engineering decision instead
of repeating the feature engineering experiments.

```python
# Create age groups
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 40, 55, 100],
    labels=["Young", "Middle-aged", "Older"]
)

# Check the distribution
df["AgeGroup"].value_counts()
```

```python
df["health_risk_count"] = (
    (df["RestingBP"] >= 140).astype(int)
    + (df["Cholesterol"] >= 240).astype(int)
    + (df["FastingBS"] == 1).astype(int)
)

df.head()
```

# Neural Network Architecture

## What is a Neural Network?

A neural network is a machine learning model composed of interconnected
neurons organized into layers.

The network learns parameters called weights and biases during training.

For a single neuron:

output = activation(dot(inputs, weights) + bias)

The neuron therefore performs three main operations:

1. Calculate a weighted sum of the inputs.
2. Add a bias.
3. Apply an activation function.

## Neural Network Layers

A neural network typically contains:

### Input Layer

Receives the input features.

For this project, the input represents the processed patient
characteristics.

### Hidden Layers

Hidden layers learn intermediate patterns and relationships from the input
data.

Adding more hidden layers allows the network to learn more complex
representations.

### Output Layer

Produces the final prediction.

For this binary classification problem, the output layer will produce the
probability of the patient belonging to the Heart Disease class.

## Planned Neural Network Architecture

The initial neural network architecture will follow a simple structure:

Input Features

        ↓

Dense Hidden Layer

        ↓

Dense Hidden Layer

        ↓

Output Layer

        ↓

Heart Disease Probability
