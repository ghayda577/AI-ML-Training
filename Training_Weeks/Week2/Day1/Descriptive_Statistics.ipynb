# Descriptive Statistics

## Step 1: Loading the Dataset Column

The dataset was loaded using `pandas.read_csv()`. 
The `encoding="ISO-8859-1"` parameter was used to correctly read special characters in the CSV file and avoid encoding errors.

```python
import numpy as np
import pandas as pd
df = pd.read_csv("Superstore.csv", encoding="ISO-8859-1")
df.head(5)

```


## Step 2: Computing Descriptive Statistics

The **Profit** column was selected because it represents the profit value for each order and can be analyzed using descriptive statistics.
Descriptive statistical measures were calculated for the Profit column, including:

- **Mean:** Represents the average profit value.
- **Median:** Represents the middle profit value after sorting the data.
- **Mode:** Represents the most frequent profit value.
- **Standard Deviation:** Measures how much the profit values vary from the mean.
- **IQR (Interquartile Range):** Measures the spread of the middle 50% of the data.


```python
profit= df.Profit

mean =profit.mean()
median =profit.median()
mode =profit.mode()
std =profit.std(ddof=0)

print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
print("Standard Deviation =", std)
```

### Interquartile Range (IQR)

The **Interquartile Range (IQR)** measures the spread of the middle **50%** of the data.

**Formula:**
\[
IQR = Q3 - Q1
\]

Where:
- **Q1 (25th percentile)** = First Quartile
- **Q3 (75th percentile)** = Third Quartile

The IQR is commonly used to detect outliers using the following boundaries:

- **Lower Bound = Q1 - 1.5 × IQR**
- **Upper Bound = Q3 + 1.5 × IQR**

```python
Q1=profit.quantile(.25)
Q2=profit.quantile(.5)
Q3=profit.quantile(.75)
IQR= Q3 - Q1
LB=Q1 - (1.5*IQR)
UB=Q3 + (1.5*IQR)
print("IQR =" , IQR)
print("Lower Bound =", LB)
print("Upper Bound =" , UB)


```

## Step 3: Choosing the Best Measure of Central Tendency

The mean profit value is approximately 28.66, while the median profit value is approximately 8.67.

Since the mean is much higher than the median and the standard deviation is large, this indicates that the Profit column contains extreme values (outliers) that affect the mean.

Therefore, the median is a better representation of the typical profit value because it is less affected by unusually high or low profit values.

## Step 4: Summary of Profit Distribution

The Profit column has an average value of approximately 28.66, with a median of approximately 8.67.

The large difference between the mean and median indicates that the data is not symmetrically distributed and contains some extreme profit values.

The standard deviation is approximately 234.25, showing that profits vary widely across different orders.
