# Probability Simulation with NumPy

This notebook demonstrates basic probability concepts using NumPy simulations, including coin flips, the normal distribution, and conditional probability.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Step 1: Simulate 10,000 Coin Flips

Generate 10,000 random coin flips using NumPy.

```python
flips = np.random.choice(["Heads","Tails"], size= 10000)
print(flips)
```

### Count the Results

Count the number of heads and tails, then calculate the proportion of heads.

```python
heads= np.sum(flips=="Heads")
tails= np.sum(flips=="Tails")
proportion = heads /10000

print("Heads =", heads)
print("Tails =", tails)
print("Proportion = " , proportion)
```

### Result

The proportion of heads is close to **0.5**, confirming that the experimental probability approaches the theoretical probability as the number of trials increases.

## Step 2: Normal Distribution

Generate 10,000 values from a normal distribution and visualize them using a histogram.

```python
data= np.random.normal(loc=50 , scale=10, size=10000)
plt.hist(data , bins=30)
plt.xlabel("Value")
plt.ylabel("frequancy")
plt.title(" Normal Distribution")
```

### Result

The histogram has a bell-shaped form centered around the mean, showing the characteristics of a normal distribution.

## Step 3: Conditional Probability

Estimate a conditional probability manually and verify it using a simulation.

### Scenario

A box contains **3 Black balls** and **2 Blue balls**.

Box = [Black, Black, Black, Blue, Blue]

Let:

- **B:** The first ball is Black.
- **A:** The second ball is Black.

We want to calculate **P(A | B)**.

---

### 1) Calculate P(B)

The probability that the first ball is Black:

P(B) = 3 / 5 = 0.6

---

### 2) Calculate P(A and B)

P(A and B) means that both the first and second balls are Black.

Probability that the first ball is Black:

P(B) = 3 / 5

Since the first Black ball is **not replaced**, there are 4 balls remaining:

[Black, Black, Blue, Blue]

Probability that the second ball is Black:

P(A | B) = 2 / 4

Therefore:

P(A and B) = (3 / 5) × (2 / 4)

P(A and B) = 6 / 20 = 0.3

---

### 3) Calculate P(A | B)

Using the conditional probability formula:

P(A | B) = P(A and B) / P(B)

P(A | B) = 0.3 / 0.6

P(A | B) = 0.5



```python

count_B = 0
count_AandB = 0

for i in range(1000):

    box=["Black", "Black", "Black", "Blue", "Blue"]
    np.random.shuffle(box)

    # B: first ball is Black
    if box[0] == "Black":
        count_B += 1

    # A and B: first and second balls are Black
    if box[0] == "Black" and box[1] == "Black":
        count_AandB += 1


p_B = count_B / 1000
p_AandB = count_AandB / 1000
p_AgivenB = p_AandB / p_B


print("P(B) =", p_B)
print("P(A and B) =", p_AandB)
print("P(A | B) =", p_AgivenB)
```

### Result

The simulated conditional probability is close to the theoretical value of **0.5**, confirming the manual calculation.
